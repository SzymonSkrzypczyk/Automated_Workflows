## Inspect your actual state

This cell only reads the current project's database. It does not perform discovery. If you changed the configured database path, update `DB` accordingly.


````python
DB = (PROJECT / 'state' / 'catalogue.sqlite').resolve()
if not DB.is_file():
    print('Run live discovery from the terminal first; no database is created here.')
else:
    with sqlite3.connect(f'{DB.as_uri()}?mode=ro', uri=True) as conn:
        conn.row_factory = sqlite3.Row
        runs = [dict(row) for row in conn.execute(
            'SELECT started_at, status, search_start, search_until, pages, sightings, new_scenes, latest_acquired, error FROM runs ORDER BY started_at, rowid')]
        print(json.dumps(runs, indent=2))
        print('Saved query watermarks:')
        for row in conn.execute('SELECT stream_key, watermark FROM streams'):
            print(dict(row))
````

## The table:

````python
import json
import subprocess
import sys
from IPython.display import Markdown, display

result = subprocess.run(
    [
        sys.executable,
        str(PROJECT / "sentinel_cli.py"),
        "report",
        "--config",
        str(PROJECT / "config" / "krakow_week.json"),
    ],
    cwd=PROJECT,
    capture_output=True,
    text=True,
    encoding="utf-8",
)

if result.returncode != 0:
    raise RuntimeError(
        f"Report failed:\n{result.stderr}\n{result.stdout}"
    )

report = json.loads(result.stdout)

report_path = PROJECT / "week_report.json"
report_path.write_text(
    json.dumps(report, indent=2, ensure_ascii=False) + "\n",
    encoding="utf-8",
)

def cell(value):
    if value is None:
        return "—"
    return str(value).replace("|", "/").replace("\n", " ")

table = [
    "| Run time (UTC) | Outcome | Search interval | Sightings | "
    "Newly discovered scenes | Newest acquisition | Observation |",
    "|---|---|---|---|---|---|---|",
]

for run in report["runs"]:
    if run["status"] == "failed":
        observation = run.get("error") or "See log"
    elif run["status"] == "running":
        observation = "No completion recorded; inspect log"
    elif run["new_scenes"] == 0:
        observation = "Completed; no new scene identities"
    else:
        observation = "Completed; new scene identities found"

    values = [
        run["started_at"],
        run["status"],
        f'{run["search_start"]} → {run["search_until"]}',
        run["sightings"],
        run["new_scenes"],
        run["latest_acquired"],
        observation,
    ]
    table.append("| " + " | ".join(map(cell, values)) + " |")

display(Markdown("\n".join(table)))

print("Saved report:", report_path)
print("Unique scenes:", report["unique_scenes"])
print("Newest acquisition:", report["latest_acquired"])
print("Newest acquisition age [hours]:", report["latest_age_hours"])
print("Saved watermark:", report["watermark"])
````
